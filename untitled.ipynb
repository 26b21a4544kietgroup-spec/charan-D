```python
# ==============================================================================
# B.TECH MATHEMATICS MINI PROJECT: BRIDGE LOAD DISTRIBUTION CALCULATOR
# Model: AX = B Matrix System with Inversion Solvers
# ==============================================================================

# ==============================================================================
# SECTION 1: IMPORTS & SETUP
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean visualization style
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.size"] = 11

# ==============================================================================
# SECTION 2: CONFIGURATION & BENCHMARK DATASET
# ==============================================================================
# Support locations along the bridge length in meters: R1, R2, R3
x_supports = np.array([0.0, 10.0, 20.0])

# Initial Load Positions (m) and Loads (kiloNewtons, kN)
# Load 1: 50 kN at 4m | Load 2 (Truck): 120 kN at 8m | Load 3: 60 kN at 15m
x_loads_initial = np.array([4.0, 8.0, 15.0])
P_initial = np.array([50.0, 120.0, 60.0])

# Moved Load Positions: Truck moves from x = 8m to x = 14m
x_loads_moved = np.array([4.0, 14.0, 15.0])
P_moved = np.array([50.0, 120.0, 60.0])

# ==============================================================================
# SECTION 3: MATHEMATICAL ENGINE (AX = B SOLVER)
# ==============================================================================
def build_coefficient_matrix(x_sup):
    """
    Constructs the 3x3 coefficient matrix A based on statics and support spacing.
    Row 1: Vertical Force Equilibrium (Sum F_y = 0)  --> [1, 1, 1]
    Row 2: Moment Equilibrium about origin (Sum M_0 = 0) --> [x1, x2, x3]
    Row 3: Structural stiffness balance constraint (Ensures det(A) != 0)
    """
    row1 = [1.0, 1.0, 1.0]
    row2 = list(x_sup)
    row3 = [1.0, -2.0, 1.0]  # Elastic stiffness influence relation

    A = np.array([row1, row2, row3], dtype=float)
    return A

def build_load_vector(x_loads, P_loads):
    """
    Constructs the 3x1 load vector B.
    B1 = Total downward force sum(P_i)
    B2 = Total moment sum(P_i * x_i)
    B3 = 0 (Internal displacement equilibrium)
    """
    total_force = np.sum(P_loads)
    total_moment = np.sum(P_loads * x_loads)

    B = np.array([total_force, total_moment, 0.0], dtype=float)
    return B

def solve_bridge_reactions(A, B):
    """
    Solves X = A^(-1) * B using matrix inversion.
    """
    det_A = np.linalg.det(A)
    if np.isclose(det_A, 0):
        raise ValueError("Matrix A is singular. Unique solution does not exist.")

    A_inv = np.linalg.inv(A)
    X = np.dot(A_inv, B)
    return A_inv, X

# Run calculations for Initial State
A = build_coefficient_matrix(x_supports)
B_init = build_load_vector(x_loads_initial, P_initial)
A_inv, R_initial = solve_bridge_reactions(A, B_init)

# Run calculations for Moved Truck State
B_moved = build_load_vector(x_loads_moved, P_moved)
_, R_moved = solve_bridge_reactions(A, B_moved)

# Calculate force change
R_delta = R_moved - R_initial

# ==============================================================================
# SECTION 4: FORMATTED OUTPUT DISPLAY
# ==============================================================================
print("=" * 65)
print("     BRIDGE LOAD DISTRIBUTION CALCULATOR (AX = B SYSTEM)     ")
print("=" * 65)

print("\n--- COEFFICIENT MATRIX (A) ---")
print(np.array2string(A, precision=2, suppress_small=True))

print("\n--- INVERSE MATRIX (A^-1) ---")
print(np.array2string(A_inv, precision=4, suppress_small=True))

print("\n" + "-" * 65)
print(f"{'Support':<10} | {'Initial R (kN)':<15} | {'Moved R (kN)':<15} | {'Change ΔR (kN)':<15}")
print("-" * 65)
for i in range(3):
    print(f"R_{i+1} ({x_supports[i]}m)   | {R_initial[i]:<15.2f} | {R_moved[i]:<15.2f} | {R_delta[i]:<15.2f}")
print("-" * 65)

# ==============================================================================
# SECTION 5: VISUALIZATION CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Bar Chart comparing reaction forces before and after movement
supports_labels = [f"R1 ({x_supports[0]}m)", f"R2 ({x_supports[1]}m)", f"R3 ({x_supports[2]}m)"]
x = np.arange(len(supports_labels))
width = 0.35

rects1 = axes[0].bar(x - width/2, R_initial, width, label='Initial State (Truck @ 8m)', color='#2b5c8f')
rects2 = axes[0].bar(x + width/2, R_moved, width, label='After Truck Moves (Truck @ 14m)', color='#d95f02')

axes[0].set_ylabel('Reaction Force (kN)', fontweight='bold')
axes[0].set_title('Support Reaction Forces Comparison', fontweight='bold')
axes[0].set_xticks(x)
axes[0].set_xticklabels(supports_labels)
axes[0].legend()
axes[0].bar_label(rects1, padding=3, fmt='%.1f')
axes[0].bar_label(rects2, padding=3, fmt='%.1f')

# Plot 2: Load redistribution change (Delta R)
colors = ['#1b9e77' if val >= 0 else '#e7298a' for val in R_delta]
rects3 = axes[1].bar(supports_labels, R_delta, color=colors, width=0.4)

axes[1].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[1].set_ylabel('Force Change ΔR (kN)', fontweight='bold')
axes[1].set_title('Net Redistribution of Load (ΔR = R_moved - R_initial)', fontweight='bold')
axes[1].bar_label(rects3, padding=3, fmt='%.1f')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 3: Mathematical Engine (Matrix Inversion)
# ==========================================

def build_coefficient_matrix(x_sup):
    """
    Constructs the 3x3 coefficient matrix A based on statics and support spacing.
    Row 1: Sum of vertical forces = 0  --> [1, 1, 1]
    Row 2: Sum of moments about x=0 = 0 --> [x1, x2, x3]
    Row 3: Structural stiffness balance constraint (Ensures det(A) != 0)
    """
    row1 = [1.0, 1.0, 1.0]
    row2 = list(x_sup)
    # Row 3 uses elastic influence balance relative to central support
    row3 = [1.0, -2.0, 1.0]

    A = np.array([row1, row2, row3], dtype=float)
    return A

def build_load_vector(x_loads, P_loads):
    """
    Constructs the 3x1 load vector B.
    B1 = Total downward force sum(P_i)
    B2 = Total moment sum(P_i * x_i)
    B3 = 0 (Internal displacement constraint balance)
    """
    total_force = np.sum(P_loads)
    total_moment = np.sum(P_loads * x_loads)

    B = np.array([total_force, total_moment, 0.0], dtype=float)
    return B

def solve_bridge_reactions(A, B):
    """
    Solves X = A^(-1) * B using explicit matrix inversion.
    """
    # Verify uniqueness of solution (determinant must be non-zero)
    det_A = np.linalg.det(A)
    if np.isclose(det_A, 0):
        raise ValueError("Matrix A is singular. Unique solution does not exist.")

    # Compute inverse matrix
    A_inv = np.linalg.inv(A)

    # Calculate reactions X = A_inv @ B
    X = np.dot(A_inv, B)
    return A_inv, X

# Compute Initial State
A = build_coefficient_matrix(x_supports)
B_init = build_load_vector(x_loads_initial, P_initial)
A_inv, R_initial = solve_bridge_reactions(A, B_init)

# Compute Moved Load State (Truck relocated)
B_moved = build_load_vector(x_loads_moved, P_moved)
_, R_moved = solve_bridge_reactions(A, B_moved)

# Calculate net change in reaction forces
R_delta = R_moved - R_initial
```

```python
# ==============================================================================
# B.TECH MATHEMATICS MINI PROJECT: ADVANCED 3D HOLOGRAPHIC ANIMATION
# Model: AX = B Matrix System with Live Deflection & HUD
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from mpl_toolkits.mplot3d import Axes3D

# ------------------------------------------------------------------------------
# SECTION 1: DARK HOLOGRAPHIC THEME SETUP
# ------------------------------------------------------------------------------
plt.style.use('dark_background')
plt.rcParams['font.family'] = 'monospace' # Gives a more "computer terminal" feel

bg_color = '#02040a' # Deep space/cyberpunk black

# ------------------------------------------------------------------------------
# SECTION 2: SYSTEM CONFIGURATION & MATH ENGINE
# ------------------------------------------------------------------------------
x_supports = np.array([0.0, 10.0, 20.0])
static_load_x = np.array([4.0, 15.0])
static_load_P = np.array([50.0, 60.0])
truck_P = 120.0

def build_coefficient_matrix(x_sup):
    return np.array([
        [1.0, 1.0, 1.0],
        list(x_sup),
        [1.0, -2.0, 1.0]
    ], dtype=float)

def build_load_vector(x_loads, P_loads):
    return np.array([
        np.sum(P_loads),
        np.sum(P_loads * x_loads),
        0.0
    ], dtype=float)

A = build_coefficient_matrix(x_supports)
A_inv = np.linalg.inv(A)

# ------------------------------------------------------------------------------
# SECTION 3: ANIMATION FIGURE SETUP
# ------------------------------------------------------------------------------
fig = plt.figure(figsize=(16, 9), facecolor=bg_color)
ax = fig.add_subplot(111, projection='3d')

frames = 120
# Truck moves back and forth for a continuous loop
truck_positions = np.concatenate([np.linspace(0, 20, frames//2), np.linspace(20, 0, frames//2)])

def draw_neon_vector(ax, x, y, z, u, v, w, color, label):
    """Creates a glowing neon effect by overlaying lines of varying thickness"""
    ax.quiver(x, y, z, u, v, w, color=color, arrow_length_ratio=0.1, linewidth=4, alpha=0.4)
    ax.quiver(x, y, z, u, v, w, color='#ffffff', arrow_length_ratio=0.1, linewidth=1.5, alpha=0.9)
    ax.text(x, y, z + w + (5 if w>0 else -10), label, color=color, fontsize=10, fontweight='bold', ha='center')

def update(frame_idx):
    ax.cla()
    ax.set_facecolor(bg_color)

    truck_x = truck_positions[frame_idx]

    # Solve System
    current_x_loads = np.append(static_load_x, truck_x)
    current_P_loads = np.append(static_load_P, truck_P)
    B = build_load_vector(current_x_loads, current_P_loads)
    R_current = np.dot(A_inv, B)

    # --- 1. DYNAMIC DECK DEFLECTION (Gaussian Curve) ---
    x_span = np.linspace(-2, 22, 60)
    y_span = np.linspace(-2, 2, 8)
    X_mesh, Y_mesh = np.meshgrid(x_span, y_span)

    # Mathematical bending based on truck position and weight
    Z_mesh = -12.0 * np.exp(-((X_mesh - truck_x)**2) / 8.0)

    ax.plot_wireframe(X_mesh, Y_mesh, Z_mesh, color='#00f0ff', alpha=0.3, linewidth=0.8)
    ax.plot([-2, 22], [0, 0], [0, 0], color='#00ffff', linewidth=2, alpha=0.5, linestyle='--')

    # --- 2. DRAW SUPPORTS & NEON REACTIONS ---
    for i, (xs, r_val) in enumerate(zip(x_supports, R_current)):
        # Support Pillars
        ax.plot([xs, xs], [0, 0], [0, -25], color='#0033ff', linewidth=6, alpha=0.5)
        ax.plot([xs, xs], [0, 0], [0, -25], color='#ffffff', linewidth=1, alpha=0.8)

        # Glowing Reaction Vectors
        draw_neon_vector(ax, xs, 0, 0, 0, 0, r_val, '#00ffff', f"R{i+1}: {r_val:.1f}kN")

    # --- 3. DRAW LOADS ---
    # Static Loads
    for x_ld, p_val in zip(static_load_x, static_load_P):
        ax.quiver(x_ld, 0, p_val, 0, 0, -p_val, color='#ffcc00', arrow_length_ratio=0.1, linewidth=2, alpha=0.6, linestyle=':')
        ax.text(x_ld, 0, p_val + 5, f"{p_val}kN", color='#ffcc00', fontsize=8, alpha=0.8, ha='center')

    # Moving Truck Load (Bright Magenta)
    draw_neon_vector(ax, truck_x, 0, truck_P, 0, 0, -truck_P, '#ff00ff', f"TRUCK\n{truck_P}kN")

    # --- 4. HOLOGRAPHIC HUD FORMATTING ---
    ax.set_xlim(-3, 23)
    ax.set_ylim(-5, 5)
    ax.set_zlim(-30, 150)

    # Hide axis ticks for a cleaner hologram look
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_zticks([])

    # Clean wireframe grid styling
    ax.xaxis.pane.fill = False
    ax.yaxis.pane.fill = False
    ax.zaxis.pane.fill = False
    ax.xaxis.pane.set_edgecolor(bg_color)
    ax.yaxis.pane.set_edgecolor(bg_color)
    ax.zaxis.pane.set_edgecolor(bg_color)

    # Live Math Readout
    hud_text = (f"SYSTEM STATUS: ACTIVE\n"
                f"MODEL: AX = B\n"
                f"TRUCK POS: {truck_x:.2f}m\n"
                f"====================\n"
                f"R1: {R_current[0]:.2f} kN\n"
                f"R2: {R_current[1]:.2f} kN\n"
                f"R3: {R_current[2]:.2f} kN")

    # Place HUD in 3D space
    ax.text2D(0.05, 0.75, hud_text, transform=ax.transAxes, color='#00ffcc', fontsize=11,
              bbox=dict(facecolor='#002233', alpha=0.5, edgecolor='#00ffcc'))

    ax.set_title("STRUCTURAL LOAD MATRIX ANALYSIS", color='#ffffff', fontsize=16, fontweight='bold', pad=10)

    # Cinematic continuous orbit
    ax.view_init(elev=20 + 5 * np.sin(frame_idx * 0.1), azim=-70 + (frame_idx * 1.5))

# ------------------------------------------------------------------------------
# SECTION 4: RENDER ANIMATION
# ------------------------------------------------------------------------------
print("Initializing Advanced Matrix Simulation...")
ani = FuncAnimation(fig, update, frames=frames, interval=40, blit=False)

plt.show()

# To save as a high-quality video for your presentation, uncomment this:
# ani.save('advanced_hologram.mp4', writer='ffmpeg', fps=30, dpi=250, bitrate=2000)
```
